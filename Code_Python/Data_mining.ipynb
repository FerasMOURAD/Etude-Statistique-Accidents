```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
import os
csv_path = '../Ressources/accidents_2024.csv' if os.path.exists('../Ressources/accidents_2024.csv') else 'Ressources/accidents_2024.csv'
data = pd.read_csv(csv_path)
print(data.head())

```

```python
data_copy = data.copy()
```

```python
data_copy = data_copy.drop(columns = ["Num_Acc", "population", "nom_dep"])
```

```python
data_copy.head()
```

```python
# L heure est deja extraite sous forme numerique (0 a 23) depuis la preparation R
if "heure" not in data_copy.columns and "hrmn" in data_copy.columns:
    data_copy["heure"] = pd.to_numeric(data_copy["hrmn"].str[:2], errors="coerce").fillna(12)
print("Distribution de la variable heure :")
print(data_copy["heure"].describe())
```

```python
# 2. On encode le département en nombre de 0 a 100 (gère proprement 2A et 2B)

data_copy['dep_code'] = LabelEncoder().fit_transform(data_copy['dep'].astype(str))
```

```python
# Selection des features X
cols_to_drop = ["an", "hrmn", "dep", "com", "lat", "long", "Nb_victimes", "saison", "date", "jour_semaine"]
X = data_copy.drop(columns=[col for col in cols_to_drop if col in data_copy.columns])
print(f"X contient {X.shape[1]} variables explicatives :", list(X.columns))
X.head()
```

```python
Y = data_copy["Nb_victimes"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size = 0.2
)
```

```python
print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)
print("y_train :", Y_train.shape)
print("y_test  :", Y_test.shape)
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

```python
#modele KNN
knn = KNeighborsClassifier(n_neighbors = 5, weights = 'distance')
```

```python
knn.fit(X_train_scaled, Y_train)
```

```python
Y_pred = knn.predict (X_test_scaled)
```

```python
Y_pred[:1000]
```

```python
comparaison  = pd.DataFrame({
    "Nb de victime reel" : Y_test.values,
    "Nb de victime predit" : Y_pred
})

comparaison.head(50)
```
